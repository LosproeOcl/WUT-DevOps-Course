# MLflow Example: Tracking Two Model Training Runs

This notebook trains a `RandomForestClassifier` on the `digits` dataset with two different hyperparameter configurations and logs each run (parameters, metrics, and model) to MLflow.

```python
%pip install mlflow scikit-learn dagshub
```

```python
import mlflow
import mlflow.sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_digits
```

```python
data = load_digits()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)
```

```python
# Optional: track runs to Dagshub. Replace the placeholders with your own repo owner/name.
import dagshub
dagshub.init(repo_owner='losproeocl', repo_name='WUT-DevOps-Course', mlflow=True)
```

```python
# Set the MLflow experiment name. It will be created if it does not exist.
experiment_name = "RandomForest_V1.0.1"
mlflow.set_experiment(experiment_name)
```

```python
def train_and_log_model(n_estimators, max_depth, run_name):
    with mlflow.start_run(run_name=run_name):
        model = RandomForestClassifier(n_estimators=n_estimators, max_depth=max_depth, random_state=42)
        model.fit(X_train, y_train)

        predictions = model.predict(X_test)
        accuracy = accuracy_score(y_test, predictions)

        mlflow.log_param("n_estimators", n_estimators)
        mlflow.log_param("max_depth", max_depth)
        mlflow.log_metric("accuracy", accuracy)
        mlflow.sklearn.log_model(model, "random_forest_model")

        print(f"Logged RandomForest model with n_estimators={n_estimators}, max_depth={max_depth}, accuracy={accuracy:.4f}")
```

```python
# Train and log Model 1
train_and_log_model(n_estimators=20, max_depth=5, run_name="random_forest_model_run_1")
```

```python
# Train and log Model 2
train_and_log_model(n_estimators=100, max_depth=10, run_name="random_forest_model_run_V1.0.2")
```
